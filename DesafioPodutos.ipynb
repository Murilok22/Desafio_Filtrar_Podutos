Descritivo do Desafio

----------------------------------

QUAIS OS 5 PRODUTOS COM MAIOR AVALIAÇÕA
QUAIS OS 5 PRODUTOS COM MENOR AVALIAÇÃO

Cálculo de Receita
Cálculo de Lucro por Unidade - subtraindo o Custo_Fabricante do Preco_Venda
Cálculo de Lucro Final - multiplicando o Lucro_Unitario pela Quantidade_Vendida

A equipe de marketing quer fazer uma campanha focada apenas em peças de PC. Filtre a base para exibir apenas o produto e categoria

Salve o DataFrame atualizado


----------------------------------

```python
import pandas as pd

df_vendas_geek = pd.read_csv("/content/vendas_geek_store.csv")

maiores_avaliacoes = df_vendas_geek.nlargest(5, 'Avaliacao_Media')[['Produto', 'Categoria', 'Avaliacao_Media']]
display(maiores_avaliacoes)

menores_avaliacoes = df_vendas_geek.nsmallest(5, 'Avaliacao_Media')[['Produto', 'Categoria', 'Avaliacao_Media']]
display(menores_avaliacoes)

df_vendas_geek.sort_values(by='Avaliacao_Media', ascending=False)

df_vendas_geek.sort_values(by='Avaliacao_Media', ascending=True)

display(df_vendas_geek[['Avaliacao_Media', 'Produto']])
```

```python
#ficha tecnica, onde exibimos os dados da tabela usando a base de dados
df_vendas_geek.info()

print("------ Analise de métricas ------")

df_vendas_geek.describe()
```

```python
# vamos criar uma nova coluna, lembrando que db e sempre funcionatio
df_vendas_geek['Lucro por Unidade'] = df_vendas_geek["Preco_Venda"] - df_vendas_geek['Custo_Fabricante']

df_vendas_geek['Lucro Final'] = df_vendas_geek["Lucro por Unidade"] * df_vendas_geek['Quantidade_Vendida']

display(df_vendas_geek)
```
